```python
import pandas as pd
import numpy as np

datos = {
    "nombre": ["Nicol", "Brenda", "Yoel", "Diego", "Elena", "Fernando"],
    "curso": ["Matematica", "Python", "R", "Sociales", "Estadistica", "Programacion"],
    "nota": [15, 17, np.nan, 14, 18, 16],
    "asistencia_pct": [90, np.nan, 85, 92, 88, 95]
}

df_original = pd.DataFrame(datos)

df_original.to_csv("estudiantes.csv", index=False)

df_original
```

```python
df = pd.read_csv("estudiantes.csv")

print("Forma del DataFrame:", df.shape)

print("\nValores faltantes por columna:")
print(df.isna().sum())
```

```python
df_limpio = df.dropna(subset=["nota"])
promedio_asistencia = df_limpio["asistencia_pct"].mean()

df_limpio["asistencia_pct"] = df_limpio["asistencia_pct"].fillna(promedio_asistencia)


print("Valores faltantes después de la limpieza:")
print(df_limpio.isna().sum())

print("\nDataFrame limpio:")
print(df_limpio)
```

```python
promedio_por_curso = df_limpio.groupby("curso")["nota"].mean()

print("Promedio de nota por curso:")
print(promedio_por_curso)
```

```python
# Guardar el DataFrame limpio
df_limpio.to_csv("estudiantes_limpio.csv", index=False)

print("Archivo estudiantes_limpio.csv guardado correctamente.")
```
