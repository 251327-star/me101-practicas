```python
import numpy as np

notas = np.array([11, 14, 20, 8, 16, 13])

print("Notas:", notas)
print("Shape:", notas.shape)
print("Tipo de dato:", notas.dtype)
print("Dimensiones:", notas.ndim)
```

```python
promedio = notas.mean()
maximo = notas.max()
cantidad_aprobados = (notas >= 10.5).sum()

print("Promedio:", promedio)
print("Máximo:", maximo)
print("Aprobados:", cantidad_aprobados)
```

```python
import pandas as pd

datos = {
    "nombre": ["Ana", "Luis", "Maria", "Carlos", "Sofia"],
    "nota": [15, 9, 18, 12, 8],
    "asistencia_pct": [90, 65, 85, 75, 80]
}

df = pd.DataFrame(datos)

print("DataFrame completo:")
print(df)

print("\nEstudiantes que cumplen los requisitos:")
filtrados = df[(df["nota"] >= 10.5) & (df["asistencia_pct"] >= 70)]

print(filtrados)
```
